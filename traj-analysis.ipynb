1. Setup 

```python
# MD trajectory analysis 
import CRISP 
import plotly.io as pio 

from CRISP.simulation_utility.atomic_indices import run_atom_indices
# 3D visualisation of atom trajectories
from CRISP.simulation_utility.atomic_traj_linemap import plot_atomic_trajectory
```

```python
# retrieving atomic indices 

file_path = './xtb_test.traj'
output_folder = './indices/' # saved as numpy arrays. give array positions for all O and H


# cutoff between oxygen atoms of h2o molecules 
custom_cutoffs = {
    ('O', 'O'): 3.0,
    ('O', 'H'): 1.2

}
# cut offs saved as .csv. which o index is paired with which H index, actual O-H bonds. 
run_atom_indices(file_path, output_folder, custom_cutoffs=custom_cutoffs)
```

2. Basic Trajectory Visualisation of one water molecule

```python
traj_file = 'xtb_test.traj'
output_dir = './atomic_traj_linemap/water_test.html'
selected_atoms=[0,1,2] # first water molecule
frame_skip=1

fig = plot_atomic_trajectory(
    traj_path=traj_file,
    indices_path=selected_atoms,
    output_dir=output_dir,
    frame_skip=frame_skip,
    plot_title='Oxygen Atom Movement Analysis',
    atom_size_scale=1.2,
    show_plot=True
)
```
